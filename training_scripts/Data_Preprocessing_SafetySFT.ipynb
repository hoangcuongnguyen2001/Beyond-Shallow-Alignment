```python
import json
import re

def strip_thinking(text):
    cleaned = re.sub(
        r'<think>.*?</think>\s*',
        '', text, flags=re.DOTALL
    )
    return cleaned.strip()

def preprocess_sft(input_path, output_path):
    processed = []

    with open(input_path, "r") as f:
        for line in f:
            row = json.loads(line.strip())

            # Strip <think> block from output
            cleaned_output = strip_thinking(row['output'])

            print(f"Skipping empty output: {cleaned_output}")


            processed.append({
                'instruction': row['instruction'],
                'input': row.get('input', ''),
                'output': cleaned_output
                # reasoning field intentionally excluded for SFT
            })

    with open(output_path, "w") as f:
        for row in processed:
            f.write(json.dumps(row) + "\n")

    print(f"Total processed: {len(processed)}")
    print(f"Saved to: {output_path}")

# Run preprocessing
preprocess_sft(
    "training_data.jsonl",
    "safety_sft_dataset.jsonl"
)

# Spot check
with open("safety_sft_dataset.jsonl") as f:
    for i, line in enumerate(f):
        if i >= 5:
            break
        row = json.loads(line)
        print(f"\nInstruction: {row['instruction'][:100]}")
        print(f"Output: {row['output'][:200]}")
        print("---")
```

```python
import json

short_output_instructions = []

with open("safety_sft_dataset.jsonl", "r") as f:
    for line in f:
        row = json.loads(line)
        if len(row['output']) < 10:
            short_output_instructions.append({
                'instruction': row['instruction'],
                'output': row['output']
            })

print(f"Found {len(short_output_instructions)} instructions with output length less than 10:")
for item in short_output_instructions:
    print(f"Instruction: {item['instruction']}")
    print(f"Output: {item['output']}")
    print("---")
```
